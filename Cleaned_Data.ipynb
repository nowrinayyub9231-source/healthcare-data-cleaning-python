## Introduction
Healthcare analytics is the systematic use of data, statistical models, and AI to extract meaningful insights from health and medical information. It helps organisations improve patient outcomes, reduce costs, and optimise operations using the patterns and trends that analytics reveals. In this domain, data quality is essential: inconsistent data leads to misleading analysis. For example, duplicate entries for the same patient can cause double billing or a fragmented patient history, and a wrong billing amount can lead to disputes that damage patient trust and the organisation’s reputation. If a discharge date comes before the admission date, the length of stay becomes negative, which makes bed occupancy and resource planning inaccurate. Mixed date formats (for example, “31 January 2024” in one column and “02/02/2024” in another) can be misread, so a patient’s stay could be calculated incorrectly.

This project uses a healthcare dataset of 55,500 records and 15 columns, where each row represents one hospital admission. It contains patient details, admission and discharge information, doctor and hospital details, insurance and billing information, and medication and test results. The objective is to clean and prepare this dataset so that it can reliably support hospital administrators in maintaining accurate patient records, reducing billing conflicts, and improving patient satisfaction, which ultimately contributes to better patient outcomes.

**Categorical Columns:** inconsistencies were checked in categorical columns using for loop.

```python
# removing inconsistencies using for loop.
titles=["Dr.", "Mr.", "Mrs.", "Miss", "Ms."]
suffixes=["Jr.", "MD", "PhD", "II", "III", "IV", "DDS", "DVM"]
for i in titles: 
    healthcare_data["Doctor"]= healthcare_data["Doctor"].str.removeprefix(i + " ")
for s in suffixes: 
     healthcare_data["Doctor"]= healthcare_data["Doctor"].str.removesuffix(" " + s)

healthcare_data["Doctor"]= healthcare_data["Doctor"].str.strip()
```

## 5. Converting Date Columns
**Problem:** Date of Admission ("Wednesday, 31 January 2024") and Discharge Date ("02/02/2024", day/month/year) were stored as text in two different formats.  
**Action:** each column was converted to datetime with its own format string.  
**Check:** no discharge date was earlier than its admission date. Admissions cover 8 May 2019 to 7 May 2024.

```python
# Convert datatype of dates in dataset to datetime().
healthcare_data["Date of Admission"]= pd.to_datetime(healthcare_data["Date of Admission"])
healthcare_data["Discharge Date"]= pd.to_datetime(healthcare_data["Discharge Date"])
healthcare_data
```

```python
# to get the max() and min() of date of admission.
print(healthcare_data["Date of Admission"].min())
print(healthcare_data["Date of Admission"].max())
```

```python
# To ensure if both discharge dates and admission dates are having same datatype.
print(healthcare_data["Date of Admission"].dtypes)
print(healthcare_data["Discharge Date"].dtypes)
```

```python
# checking if any records are having any discharge date leading than admission date.
invalid_dates=(healthcare_data["Discharge Date"] < healthcare_data["Date of Admission"]).sum()
print("invalid dates:", invalid_dates)
```

## 6. Billing Amount: Negative Values and Rounding
**Problem:** 106 billing amounts were negative, ranging from about -24 to -2,008, whereas typical bills are around 25,000.  
**Action:** they were converted to positive with `.abs()`.  
**Reason:** a hospital bill cannot be negative, and the small size and low number of these values suggest sign errors in data entry.  

*Assumption: if the source confirms they are refunds, they should be handled separately.* The amounts were then rounded to 2 decimal places, the standard format for currency.

```python
# Counting the negative records.
(healthcare_data["Billing Amount"]<0).sum()
```

```python
# to convert the negative values to positive values, 
# A billing amount on an admission record should be positive. A negative total doesn’t make sense for a patient’s hospital bill.
healthcare_data["Billing Amount"]=healthcare_data["Billing Amount"].abs()
healthcare_data["Billing Amount"].head()
```

```python
# to verify there are no longer any negative values in billing amount.
healthcare_data["Billing Amount"].min()
```

```python
# dataframe shape after cleaning.
healthcare_data.shape
```

```python
# Rounding off the decimal place of billing amounts into 2.
healthcare_data["Billing Amount"]=healthcare_data["Billing Amount"].round(2)
healthcare_data.head()
```

## Adding New Columns
Two new columns, **Month of Admission** and **Year of Admission**, were created from the Date of Admission column using the pandas `.dt` accessor (`.dt.month_name()` and `.dt.year`). This was possible because the date column had already been converted to datetime format. Splitting the date into separate month and year values makes it easier to group and compare records over time in later analysis, for example to look at monthly admission patterns, yearly trends in billing, or seasonal changes in admission types. The admission dates span 2019 to 2024, so the new columns cover six calendar years.

```python
# adding 2 columns to dataframe--month of admission, year of admission
healthcare_data["Month of Admission"]=healthcare_data["Date of Admission"].dt.month_name()
```

```python
healthcare_data["Year of Admission"]=healthcare_data["Date of Admission"].dt.year
```
